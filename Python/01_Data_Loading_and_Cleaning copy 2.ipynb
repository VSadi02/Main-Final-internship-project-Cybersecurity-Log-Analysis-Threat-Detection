```python
#Check Missing Values

# df.isnull().sum()


#To calculate the percentage of missing values:
missing = (df.isnull().sum() / len(df)) * 100

missing.sort_values(ascending=False)
```

```python
#Check Duplicate Records
df.duplicated().sum()

# If duplicates exist:

df = df.drop_duplicates()

# # Verify:

df.duplicated().sum()
```

```python
# Convert Login Timestamp
df["Login_Timestamp"] = pd.to_datetime(
    df["Login_Timestamp"],
    errors="coerce"                  
)                                              #errors="coerce" converts invalid timestamps into NaT (Not a Time) instead of stopping with an error.
```
